```python
# Import Required Libraries
import pandas as pd
import numpy as np

# Ignore Warnings
import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
```

```python
# Load dataset
df = pd.read_csv("../dataset/processed/airline_delay_selected.csv",encoding="latin1")
```

```python
# Display the first five rows
df.head()
```

```python
#Shape of the dataset
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")
```

```python
# Missing Values
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": round(df.isnull().mean()*100,2)
})

missing = missing[missing["Missing Values"]>0]
missing.sort_values("Percentage (%)", ascending=False)
```

```python
#Duplicate Rows
duplicate_rows = df[df.duplicated()]
duplicate_rows_count = duplicate_rows.shape[0]
print(f"Duplicate Rows : {duplicate_rows_count}")
```

```python
#Data Types
df.info()
```

```python
# Statistical Summary of Numerical Columns
df.describe().T
```

```python
# Statistical Summary of Categorical Columns
df.describe(include="object").T
```

```python
# Drop rows with missing OriginState or DestState
df.dropna(subset=["OriginState", "DestState"], inplace=True)
```

```python
delay_columns = [
    "CarrierDelay",
    "WeatherDelay",
    "NASDelay",
    "SecurityDelay",
    "LateAircraftDelay"
]

df[delay_columns] = df[delay_columns].fillna(0)
```

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": round(df.isnull().mean() * 100, 2)
})

missing = missing[missing["Missing Values"] > 0]
missing.sort_values("Percentage (%)", ascending=False)
```

```python
# Convert FlightDate to datetime
df["FlightDate"] = pd.to_datetime(df["FlightDate"])
```

```python
# Convert categorical columns to category dtype
categorical_columns = [
    "Reporting_Airline",
    "Origin",
    "OriginCityName",
    "OriginState",
    "Dest",
    "DestCityName",
    "DestState",
    "CancellationCode"
]

for col in categorical_columns:
    df[col] = df[col].astype("category")
```

```python
# Dataset information
df.info()

# Check for remaining duplicates
print("Duplicate Records:", df.duplicated().sum())

# Verify missing values
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage (%)": round(df.isnull().mean() * 100, 2)
})

missing[missing["Missing Values"] > 0]
```

```python
df.to_csv(
    "../dataset/processed/airline_delay_cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```

# Key Observations

- Missing values in delay reason columns were replaced with `0` as they indicate that the corresponding delay did not occur.
- Rows with missing origin or destination state information were removed due to their negligible impact on the dataset.
- Remaining missing values were retained because they represent valid operational scenarios such as cancelled or diverted flights.
- The `FlightDate` column was converted to a datetime format, and categorical columns were optimized to improve memory efficiency.
- The cleaned dataset is now ready for exploratory data analysis.
